# Chicken Threat Detector Data v4

This notebook is the **data-collection and dataset-build stage** for the v4 chicken threat detector. It keeps the v3 YOLO pipeline structure, then adds camera-trap predator sources and explicit day/night balancing.

Pipeline boundary:

- Download / import raw YOLO sources.
- Convert polygon labels to detection boxes where needed.
- Map source classes into the canonical detector classes.
- Add high-value camera-trap predator sources from LILA / COCO Camera Traps metadata.
- Download only selected camera-trap images instead of giant archives.
- Preserve source, location, timestamp, and inferred day/night metadata.
- Balance the final dataset using v4 class targets and the requested +10% day/night predator targets.
- Validate and visually inspect the final YOLO dataset.
- Export `chicken_threat_dataset_v4.zip`, which is the input for the v4 model notebook.

## Shared YOLO helpers

```python
@dataclass(frozen=True)
class ProcessStats:
    source: str
    images_seen: int = 0
    images_written: int = 0
    labels_seen: int = 0
    labels_written: int = 0
    empty_label_images: int = 0
    missing_label_images: int = 0
    dropped_labels: int = 0
    invalid_labels: int = 0
    skipped_no_box: int = 0
    skipped_download: int = 0


def clean_dir(path: Path) -> None:
    path = path.resolve()
    allowed_roots = [WORK_DIR.resolve(), PROCESSED_ROOT.resolve(), FINAL_DATASET_ROOT.resolve()]
    if not any(path == root or root in path.parents for root in allowed_roots):
        raise ValueError(f"Refusing to clean path outside the notebook work directory: {path}")
    if path.exists():
        shutil.rmtree(path)
    path.mkdir(parents=True, exist_ok=True)


def slugify(value: str) -> str:
    value = re.sub(r"[^A-Za-z0-9_.-]+", "_", value.strip())
    return value.strip("._") or "item"


def load_class_names(dataset_root: Path) -> list[str]:
    data_yaml = dataset_root / "data.yaml"
    if not data_yaml.exists():
        raise FileNotFoundError(f"Missing data.yaml in {dataset_root}")

    data = yaml.safe_load(data_yaml.read_text(encoding="utf-8"))
    names = data.get("names")
    if isinstance(names, dict):
        return [str(names[key]) for key in sorted(names, key=lambda key: int(key))]
    if isinstance(names, list):
        return [str(name) for name in names]
    raise ValueError(f"Unsupported names format in {data_yaml}")


def write_data_yaml(dataset_root: Path, names: list[str] | None = None) -> None:
    names = list(names or CANONICAL_NAMES)
    data = {
        "path": str(dataset_root.resolve()),
        "train": "train/images",
        "val": "valid/images",
        "test": "test/images",
        "nc": len(names),
        "names": names,
    }
    (dataset_root / "data.yaml").write_text(yaml.safe_dump(data, sort_keys=False), encoding="utf-8")

def image_files(image_dir: Path) -> list[Path]:
    if not image_dir.exists():
        return []
    return sorted(path for path in image_dir.iterdir() if path.suffix.lower() in IMAGE_SUFFIXES)


def copy_image(source: Path, dest_dir: Path, dest_stem: str) -> Path:
    dest_dir.mkdir(parents=True, exist_ok=True)
    dest = dest_dir / f"{dest_stem}{source.suffix.lower()}"
    shutil.copy2(source, dest)
    return dest


def polygon_to_box(coords: list[float]) -> tuple[float, float, float, float] | None:
    if len(coords) < 6 or len(coords) % 2 != 0:
        return None
    xs = coords[0::2]
    ys = coords[1::2]
    x_min, x_max = max(0.0, min(xs)), min(1.0, max(xs))
    y_min, y_max = max(0.0, min(ys)), min(1.0, max(ys))
    width = x_max - x_min
    height = y_max - y_min
    if width <= 0 or height <= 0:
        return None
    return ((x_min + x_max) / 2, (y_min + y_max) / 2, width, height)


def clip_yolo_box(box: Iterable[float]) -> tuple[float, float, float, float] | None:
    x_center, y_center, width, height = [float(value) for value in box]
    if width <= 0 or height <= 0:
        return None

    x_min = max(0.0, x_center - width / 2)
    y_min = max(0.0, y_center - height / 2)
    x_max = min(1.0, x_center + width / 2)
    y_max = min(1.0, y_center + height / 2)
    width = x_max - x_min
    height = y_max - y_min
    if width <= 0 or height <= 0:
        return None
    return ((x_min + x_max) / 2, (y_min + y_max) / 2, width, height)


def parse_yolo_line(line: str) -> tuple[int, tuple[float, float, float, float]] | None:
    parts = line.split()
    if len(parts) < 5:
        return None
    try:
        class_id = int(float(parts[0]))
        coords = [float(value) for value in parts[1:]]
    except ValueError:
        return None

    if len(coords) == 4:
        box = clip_yolo_box(coords)
    else:
        box = polygon_to_box(coords)
    if box is None:
        return None
    return class_id, box


def format_yolo_line(class_id: int, box: tuple[float, float, float, float]) -> str:
    return f"{class_id} " + " ".join(f"{value:.6f}" for value in box)
```

## Inspect source datasets

```python
def summarize_yolo_source(dataset_root: Path) -> dict[str, Counter]:
    class_names = load_class_names(dataset_root)
    summary: dict[str, Counter] = {}

    for split in SPLITS:
        label_dir = dataset_root / split / "labels"
        image_dir = dataset_root / split / "images"
        counter: Counter = Counter()
        counter["__images__"] = len(image_files(image_dir))
        counter["__empty_label_files__"] = 0
        counter["__missing_label_files__"] = 0

        for image_path in image_files(image_dir):
            label_path = label_dir / f"{image_path.stem}.txt"
            if not label_path.exists():
                counter["__missing_label_files__"] += 1
                continue
            text = label_path.read_text(encoding="utf-8").strip()
            if not text:
                counter["__empty_label_files__"] += 1
                continue
            for line in text.splitlines():
                parsed = parse_yolo_line(line)
                if parsed is None:
                    counter["__invalid_labels__"] += 1
                    continue
                class_id, _ = parsed
                class_name = class_names[class_id] if 0 <= class_id < len(class_names) else f"unknown:{class_id}"
                counter[class_name] += 1

        summary[split] = counter
    return summary


def print_summary(summary: dict[str, Counter]) -> None:
    for split, counter in summary.items():
        object_total = sum(count for label, count in counter.items() if not label.startswith("__"))
        print(f"{split.upper()} ({object_total} objects, {counter['__images__']} images)")
        for label, count in counter.most_common():
            if label.startswith("__"):
                continue
            print(f"  - {label}: {count}")
        print(f"  empty label files: {counter['__empty_label_files__']}")
        print(f"  missing label files: {counter['__missing_label_files__']}")
        if counter["__invalid_labels__"]:
            print(f"  invalid labels: {counter['__invalid_labels__']}")
        print()


if ROBOFLOW_RAW_ROOT.exists():
    print_summary(summarize_yolo_source(ROBOFLOW_RAW_ROOT))
else:
    print(f"Roboflow raw dataset not found yet: {ROBOFLOW_RAW_ROOT}")
```

## Convert and filter YOLO sources

`process_yolo_source` converts segmentation polygons to detection boxes, drops unwanted labels, rewrites class IDs into the canonical class map, and copies only usable images into a processed source folder.


```python
def process_yolo_source(
    source_root: Path,
    output_root: Path,
    class_map: dict[str, str | None],
    source_name: str,
    keep_empty_images: bool = False,
) -> ProcessStats:
    source_root = Path(source_root)
    output_root = Path(output_root)
    class_names = load_class_names(source_root)
    clean_dir(output_root)

    totals = defaultdict(int)

    for split in SPLITS:
        src_image_dir = source_root / split / "images"
        src_label_dir = source_root / split / "labels"
        dst_image_dir = output_root / split / "images"
        dst_label_dir = output_root / split / "labels"
        dst_image_dir.mkdir(parents=True, exist_ok=True)
        dst_label_dir.mkdir(parents=True, exist_ok=True)

        for image_path in tqdm(image_files(src_image_dir), desc=f"{source_name}:{split}"):
            totals["images_seen"] += 1
            label_path = src_label_dir / f"{image_path.stem}.txt"
            if not label_path.exists():
                totals["missing_label_images"] += 1
                if not keep_empty_images:
                    continue
                lines = []
            else:
                text = label_path.read_text(encoding="utf-8").strip()
                if not text:
                    totals["empty_label_images"] += 1
                    if not keep_empty_images:
                        continue
                    lines = []
                else:
                    lines = text.splitlines()

            output_lines: list[str] = []
            for line in lines:
                parsed = parse_yolo_line(line)
                if parsed is None:
                    totals["invalid_labels"] += 1
                    continue
                old_class_id, box = parsed
                totals["labels_seen"] += 1
                if not 0 <= old_class_id < len(class_names):
                    totals["invalid_labels"] += 1
                    continue

                source_class = class_names[old_class_id]
                canonical_class = class_map.get(source_class)
                if canonical_class is None:
                    totals["dropped_labels"] += 1
                    continue
                if canonical_class not in CANONICAL_ID:
                    raise ValueError(f"Unknown canonical class {canonical_class!r} for {source_class!r}")

                output_lines.append(format_yolo_line(CANONICAL_ID[canonical_class], box))

            if not output_lines and not keep_empty_images:
                continue

            dest_stem = f"{slugify(source_name)}_{slugify(image_path.stem)}"
            copy_image(image_path, dst_image_dir, dest_stem)
            (dst_label_dir / f"{dest_stem}.txt").write_text("\n".join(output_lines) + ("\n" if output_lines else ""), encoding="utf-8")
            totals["images_written"] += 1
            totals["labels_written"] += len(output_lines)

    write_data_yaml(output_root)
    return ProcessStats(source=source_name, **totals)


def process_poultry_dataset() -> Path | None:
    if not ROBOFLOW_RAW_ROOT.exists():
        print(f"Poultry raw dataset not found: {ROBOFLOW_RAW_ROOT}")
        return None

    output_root = PROCESSED_ROOT / "poultry"
    stats = process_yolo_source(
        source_root=ROBOFLOW_RAW_ROOT,
        output_root=output_root,
        class_map=POULTRY_CLASS_MAP,
        source_name="poultry",
        keep_empty_images=False,
    )
    print(stats)
    print_summary(summarize_yolo_source(output_root))
    return output_root


poultry_processed_root = process_poultry_dataset()
```

## Download and export balanced Open Images predator/context samples

Open Images is downloaded per source class instead of with one global `max_samples`.
This avoids collecting thousands of `Person` instances while still giving fox, cat, dog, and bird-of-prey enough samples.


```python
def find_fiftyone_detections_field(dataset) -> str:
    for field_name, field in dataset.get_field_schema().items():
        document_type = getattr(field, "document_type", None)
        if document_type is not None and document_type.__name__ == "Detections":
            return field_name
    raise ValueError("Could not find a FiftyOne Detections field in the dataset")


def export_fiftyone_split_to_yolo(
    dataset,
    output_root: Path,
    split: str,
    class_map: dict[str, str | None],
    source_name: str,
) -> ProcessStats:
    detection_field = find_fiftyone_detections_field(dataset)
    dst_image_dir = output_root / split / "images"
    dst_label_dir = output_root / split / "labels"
    dst_image_dir.mkdir(parents=True, exist_ok=True)
    dst_label_dir.mkdir(parents=True, exist_ok=True)

    totals = defaultdict(int)

    for sample in tqdm(dataset, desc=f"{source_name}:{split}"):
        totals["images_seen"] += 1
        detections = sample[detection_field].detections if sample[detection_field] is not None else []
        output_lines: list[str] = []

        for detection in detections:
            totals["labels_seen"] += 1
            canonical_class = class_map.get(detection.label)
            if canonical_class is None:
                totals["dropped_labels"] += 1
                continue
            if canonical_class not in CANONICAL_ID:
                raise ValueError(f"Unknown canonical class {canonical_class!r} for {detection.label!r}")

            x_min, y_min, width, height = [float(value) for value in detection.bounding_box]
            box = clip_yolo_box((x_min + width / 2, y_min + height / 2, width, height))
            if box is None:
                totals["invalid_labels"] += 1
                continue
            output_lines.append(format_yolo_line(CANONICAL_ID[canonical_class], box))

        if not output_lines:
            continue

        filepath = Path(sample.filepath)
        dest_stem = f"{slugify(source_name)}_{slugify(split)}_{slugify(sample.id)}"
        copy_image(filepath, dst_image_dir, dest_stem)
        (dst_label_dir / f"{dest_stem}.txt").write_text("\n".join(output_lines) + "\n", encoding="utf-8")
        totals["images_written"] += 1
        totals["labels_written"] += len(output_lines)

    return ProcessStats(source=f"{source_name}:{split}", **totals)


def build_open_images_source(output_root: Path = PROCESSED_ROOT / "open_images_balanced") -> Path:
    import fiftyone.zoo as foz

    clean_dir(output_root)
    stats: list[ProcessStats] = []
    split_map = {"train": "train", "valid": "validation", "test": "test"}

    for canonical_split, open_images_split in split_map.items():
        targets = OPEN_IMAGES_TARGET_IMAGES_BY_SOURCE_CLASS[canonical_split]

        for source_class, max_samples in targets.items():
            if max_samples <= 0:
                continue
            if source_class not in OPEN_IMAGES_CLASS_MAP:
                raise ValueError(f"{source_class!r} has no OPEN_IMAGES_CLASS_MAP entry")

            source_map = {source_class: OPEN_IMAGES_CLASS_MAP[source_class]}
            dataset_name = f"chicken-threat-open-images-{open_images_split}-{slugify(source_class)}"

            dataset = foz.load_zoo_dataset(
                "open-images-v7",
                split=open_images_split,
                label_types=["detections"],
                classes=[source_class],
                max_samples=max_samples,
                shuffle=True,
                seed=42,
                # Only export the target class for this shard. This prevents Person from leaking
                # into every animal subset and dominating the final distribution.
                only_matching=True,
                dataset_name=dataset_name,
                overwrite=True,
            )
            stats.append(
                export_fiftyone_split_to_yolo(
                    dataset=dataset,
                    output_root=output_root,
                    split=canonical_split,
                    class_map=source_map,
                    source_name=f"open_images_{source_class}",
                )
            )

    write_data_yaml(output_root)
    for item in stats:
        print(item)
    print_summary(summarize_yolo_source(output_root))
    return output_root


if DOWNLOAD_OPEN_IMAGES:
    open_images_processed_root = build_open_images_source()
else:
    open_images_processed_root = PROCESSED_ROOT / "open_images_balanced" if (PROCESSED_ROOT / "open_images_balanced").exists() else None
    print("Skipping Open Images download. Set DOWNLOAD_OPEN_IMAGES = True to run it.")
```

## Download and convert LILA camera-trap sources

These sources use COCO Camera Traps or COCO-style metadata. The notebook downloads metadata first, samples matching classes, then downloads only selected image files from the public cloud folder.

UNSW Predators is included in configuration because it is high-value fox/dingo/quoll context, but it is image-level CCT metadata. Provide `megadetector_results_path` to assign animal boxes from MegaDetector output; otherwise it is reported and skipped rather than using unsafe full-image boxes.

```python
def normalize_label(value: str | None) -> str:
    return re.sub(r"\s+", " ", str(value or "").replace("_", " ").strip().lower())


def cache_path_for_url(url: str, cache_dir: Path) -> Path:
    parsed = urlparse(url)
    name = Path(parsed.path).name or hashlib.sha1(url.encode("utf-8")).hexdigest()
    return cache_dir / name


def download_url(url: str, destination: Path, overwrite: bool = False) -> Path:
    destination.parent.mkdir(parents=True, exist_ok=True)
    if destination.exists() and destination.stat().st_size > 0 and not overwrite:
        return destination
    with requests.get(url, stream=True, timeout=60) as response:
        response.raise_for_status()
        with destination.open("wb") as file:
            for chunk in response.iter_content(chunk_size=1024 * 1024):
                if chunk:
                    file.write(chunk)
    return destination


def load_json_resource(path_or_url: str, cache_dir: Path) -> dict:
    cache_dir.mkdir(parents=True, exist_ok=True)
    if str(path_or_url).startswith(("http://", "https://")):
        path = download_url(path_or_url, cache_path_for_url(path_or_url, cache_dir))
    else:
        path = Path(path_or_url).expanduser().resolve()

    if path.suffix.lower() == ".zip":
        with zipfile.ZipFile(path) as archive:
            json_names = [name for name in archive.namelist() if name.lower().endswith(".json")]
            if not json_names:
                raise FileNotFoundError(f"No JSON file found in {path}")
            selected = sorted(json_names, key=lambda name: (len(name), name))[0]
            extracted = cache_dir / Path(selected).name
            if not extracted.exists():
                with archive.open(selected) as source_file, extracted.open("wb") as dest_file:
                    shutil.copyfileobj(source_file, dest_file)
            path = extracted

    return json.loads(path.read_text(encoding="utf-8"))


def category_name(category: dict | str | None) -> str:
    if isinstance(category, str):
        return category
    if not isinstance(category, dict):
        return ""
    for key in ("name", "common_name", "label", "id"):
        if category.get(key) is not None:
            return str(category[key])
    return ""


def annotation_category_name(annotation: dict, categories_by_id: dict[str, str]) -> str:
    for key in ("category", "category_name", "label", "name", "species"):
        if annotation.get(key) is not None:
            return str(annotation[key])
    return categories_by_id.get(str(annotation.get("category_id")), str(annotation.get("category_id", "")))


def image_file_name(image_info: dict) -> str:
    for key in ("file_name", "file", "filepath", "path", "id"):
        if image_info.get(key):
            return str(image_info[key]).replace("\\", "/")
    raise ValueError(f"Image record has no file path field: {image_info}")


def build_image_url(image_base_url: str, file_name: str) -> str:
    return image_base_url.rstrip("/") + "/" + quote(file_name.lstrip("/"), safe="/()_.-~")


def stable_split(key: str, location: str | None = None) -> str:
    split_key = location or key
    value = int(hashlib.sha1(str(split_key).encode("utf-8")).hexdigest()[:8], 16) / 0xFFFFFFFF
    if value < 0.80:
        return "train"
    if value < 0.90:
        return "valid"
    return "test"


def infer_night_from_text(value: str | None) -> bool | None:
    if not value:
        return None
    match = re.search(r"(?:^|[^0-9])([01][0-9]|2[0-3])[-_:hH]?([0-5][0-9])", str(value))
    if not match:
        return None
    hour = int(match.group(1))
    return hour < 6 or hour >= 20


def infer_is_night(image_info: dict, image_path: Path | None = None) -> bool:
    for key in ("is_night", "night", "infrared", "ir", "flash"):
        value = image_info.get(key)
        if isinstance(value, bool):
            return value
        if str(value).strip().lower() in {"true", "1", "yes", "night", "ir", "infrared"}:
            return True

    for key in ("datetime", "date_time", "date_captured", "timestamp", "file_name", "id"):
        parsed = infer_night_from_text(str(image_info.get(key, "")))
        if parsed is not None:
            return parsed

    return False


def coco_bbox_to_yolo(
    bbox: Iterable[float],
    image_width: int,
    image_height: int,
) -> tuple[float, float, float, float] | None:
    values = [float(value) for value in bbox]
    if len(values) != 4:
        return None

    x, y, width, height = values
    if width <= 0 or height <= 0:
        return None

    if max(abs(x), abs(y), abs(width), abs(height)) <= 1.0:
        return clip_yolo_box((x + width / 2, y + height / 2, width, height))

    if image_width <= 0 or image_height <= 0:
        return None

    return clip_yolo_box((
        (x + width / 2) / image_width,
        (y + height / 2) / image_height,
        width / image_width,
        height / image_height,
    ))


def largest_megadetector_animal_box(md_image: dict | None) -> list[float] | None:
    if not md_image:
        return None

    best = None
    best_area = 0.0
    for detection in md_image.get("detections", []):
        if str(detection.get("category")) != "1":
            continue
        if float(detection.get("conf", detection.get("confidence", 0.0))) < 0.35:
            continue

        bbox = detection.get("bbox")
        if not bbox or len(bbox) != 4:
            continue

        area = float(bbox[2]) * float(bbox[3])
        if area > best_area:
            best = bbox
            best_area = area

    return best


def source_download_caps(source: dict, split: str) -> dict[str, int]:
    source_caps = source.get("object_caps_by_split", {}).get(split)
    if source_caps:
        return source_caps
    return LILA_SOURCE_OBJECT_CAPS_BY_SPLIT[split]


def select_cct_image_items(
    source: dict,
    image_items: list[tuple[str, list[dict]]],
    images_by_id: dict[str, dict],
) -> list[tuple[str, list[dict]]]:
    planned_counts_by_split_class = {split: Counter() for split in SPLITS}
    planned_images_by_split = Counter()
    selected_image_items = []

    max_images_total = source.get("max_download_images")
    max_images_by_split = source.get("max_download_images_by_split", {})

    for image_id, annotations in image_items:
        image_info = images_by_id.get(str(image_id), {"id": image_id, "file_name": image_id})
        file_name = image_file_name(image_info)
        location = str(image_info.get("location", image_info.get("location_id", image_info.get("seq_id", ""))) or "")
        split = stable_split(file_name, location or None)
        caps = source_download_caps(source, split)
        image_counts = Counter(annotation["canonical_class"] for annotation in annotations)

        capped_classes = {
            name
            for name in image_counts
            if name in caps and caps[name] > 0
        }
        if not capped_classes:
            continue

        if max_images_total is not None and len(selected_image_items) >= max_images_total:
            break

        split_image_cap = max_images_by_split.get(split)
        if split_image_cap is not None and planned_images_by_split[split] >= split_image_cap:
            continue

        useful = any(
            planned_counts_by_split_class[split][name] < caps[name]
            for name in capped_classes
        )
        if not useful:
            continue

        selected_image_items.append((image_id, annotations))
        planned_images_by_split[split] += 1

        for name, value in image_counts.items():
            if name in caps:
                planned_counts_by_split_class[split][name] += value

    print("Planned capped downloads:")
    for split in SPLITS:
        print(f"  {split}: {planned_images_by_split[split]} images")
        for name, count in planned_counts_by_split_class[split].most_common():
            cap = source_download_caps(source, split).get(name)
            if cap is not None:
                print(f"    - {name}: planned {count} / cap {cap}")

    return selected_image_items


def process_cct_source(source: dict, output_root: Path | None = None) -> Path | None:
    if not source.get("enabled", True):
        print(f"Skipping disabled source: {source['name']}")
        return None

    source_name = source["name"]
    output_root = output_root or PROCESSED_ROOT / slugify(source_name)
    clean_dir(output_root)

    for split in SPLITS:
        (output_root / split / "images").mkdir(parents=True, exist_ok=True)
        (output_root / split / "labels").mkdir(parents=True, exist_ok=True)

    metadata_ref = source.get("metadata_url") or source.get("metadata_path")
    if not metadata_ref:
        raise ValueError(f"Source {source_name} has no metadata_url or metadata_path")

    data = load_json_resource(metadata_ref, CACHE_ROOT / slugify(source_name))
    categories_by_id = {
        str(category.get("id")): category_name(category)
        for category in data.get("categories", [])
    }
    images_by_id = {
        str(image.get("id", image_file_name(image))): image
        for image in data.get("images", [])
    }

    md_by_file: dict[str, dict] = {}
    if source.get("megadetector_results_path"):
        md_data = load_json_resource(
            source["megadetector_results_path"],
            CACHE_ROOT / f"{slugify(source_name)}_md",
        )
        md_by_file = {
            str(item.get("file", item.get("file_name", ""))).replace("\\", "/"): item
            for item in md_data.get("images", [])
        }

    annotations_by_image: dict[str, list[dict]] = defaultdict(list)
    totals = defaultdict(int)

    for annotation in data.get("annotations", []):
        totals["labels_seen"] += 1
        image_id = str(annotation.get("image_id", annotation.get("image", annotation.get("id", ""))))
        source_class = normalize_label(annotation_category_name(annotation, categories_by_id))
        canonical = source.get("class_map", {}).get(source_class)

        if canonical is None:
            totals["dropped_labels"] += 1
            continue

        has_box = bool(annotation.get("bbox"))
        if not has_box and not md_by_file:
            totals["skipped_no_box"] += 1
            continue

        item = dict(annotation)
        item["canonical_class"] = canonical
        annotations_by_image[image_id].append(item)

    if not annotations_by_image:
        print(f"No usable boxed annotations for {source_name}. Add megadetector_results_path for image-level CCT sources.")
        write_data_yaml(output_root)
        return output_root

    image_items = list(annotations_by_image.items())
    random.Random(42).shuffle(image_items)

    selected_image_items = select_cct_image_items(
        source=source,
        image_items=image_items,
        images_by_id=images_by_id,
    )

    print(f"{source_name}: selected {len(selected_image_items)} images from {len(image_items)} candidates")

    counts_by_split_class = {split: Counter() for split in SPLITS}
    manifest_rows = []

    for image_id, annotations in tqdm(selected_image_items, desc=source_name):
        image_info = images_by_id.get(str(image_id), {"id": image_id, "file_name": image_id})
        file_name = image_file_name(image_info)
        location = str(image_info.get("location", image_info.get("location_id", image_info.get("seq_id", ""))) or "")
        split = stable_split(file_name, location or None)
        image_counts = Counter(annotation["canonical_class"] for annotation in annotations)
        caps = source_download_caps(source, split)

        if not any(counts_by_split_class[split][name] < caps.get(name, 0) for name in image_counts):
            continue

        image_url = (
            build_image_url(source["image_base_url"], file_name)
            if source.get("image_base_url")
            else str(Path(source["image_root"]) / file_name)
        )
        suffix = Path(file_name).suffix.lower() if Path(file_name).suffix.lower() in IMAGE_SUFFIXES else ".jpg"
        dest_stem = f"{slugify(source_name)}_{slugify(Path(file_name).with_suffix('').as_posix())}"
        dest_image = output_root / split / "images" / f"{dest_stem}{suffix}"

        try:
            if source.get("image_root"):
                local_image = Path(source["image_root"]) / file_name
                if not local_image.exists():
                    raise FileNotFoundError(local_image)
                shutil.copy2(local_image, dest_image)
            else:
                download_url(image_url, dest_image)

            with Image.open(dest_image) as image:
                image_width, image_height = image.size

        except Exception as exc:
            totals["skipped_download"] += 1
            if dest_image.exists():
                dest_image.unlink()
            if totals["skipped_download"] <= 10:
                print(f"Skipping failed image load {image_url}: {exc}")
            continue

        md_image = md_by_file.get(file_name) or md_by_file.get(file_name.lstrip("/"))
        output_lines = []
        written_counts = Counter()

        for annotation in annotations:
            bbox = annotation.get("bbox") or largest_megadetector_animal_box(md_image)
            if not bbox:
                totals["skipped_no_box"] += 1
                continue

            box = coco_bbox_to_yolo(bbox, image_width, image_height)
            if box is None:
                totals["invalid_labels"] += 1
                continue

            canonical = annotation["canonical_class"]
            output_lines.append(format_yolo_line(CANONICAL_ID[canonical], box))
            written_counts[canonical] += 1

        if not output_lines:
            dest_image.unlink(missing_ok=True)
            continue

        label_path = output_root / split / "labels" / f"{dest_stem}.txt"
        label_path.write_text("\n".join(output_lines) + "\n", encoding="utf-8")

        counts_by_split_class[split].update(written_counts)
        totals["images_seen"] += 1
        totals["images_written"] += 1
        totals["labels_written"] += sum(written_counts.values())

        manifest_rows.append({
            "image": str(dest_image.relative_to(output_root)).replace("\\", "/"),
            "label": str(label_path.relative_to(output_root)).replace("\\", "/"),
            "source": source_name,
            "source_file": file_name,
            "source_url": image_url,
            "split": split,
            "condition": "night" if infer_is_night(image_info, dest_image) else "day",
            "location": location,
            "counts": dict(written_counts),
        })

    with (output_root / "source_manifest_v4.jsonl").open("w", encoding="utf-8") as manifest_file:
        for row in manifest_rows:
            manifest_file.write(json.dumps(row, sort_keys=True) + "\n")

    write_data_yaml(output_root)
    print(ProcessStats(source=source_name, **totals))
    print_summary(summarize_yolo_source(output_root))
    return output_root


lila_processed_roots: list[Path] = []
if DOWNLOAD_LILA_CAMERA_TRAPS:
    for source in [*LILA_CCT_SOURCES, *LOCAL_COCO_SOURCES]:
        root = process_cct_source(source)
        if root is not None and Path(root).exists():
            lila_processed_roots.append(Path(root))
else:
    print("Skipping LILA camera-trap sources. Set DOWNLOAD_LILA_CAMERA_TRAPS = True to run them.")
```

```python
BARN_ROBOFLOW_WORKSPACE = "paul-thumfart"
BARN_ROBOFLOW_PROJECT = "chicken-barn-images"
BARN_ROBOFLOW_VERSION = None
BARN_DATASET_ZIP_PATH = None  # Optional local path if not using Colab upload.


def find_yolo_dataset_root(root: Path) -> Path:
    root = Path(root).resolve()
    candidates = [root, *[path.parent for path in root.rglob("data.yaml")]]
    for candidate in candidates:
        if (candidate / "data.yaml").exists() and all(
            (candidate / split / "images").exists()
            for split in ("train", "valid")
        ):
            return candidate
    raise FileNotFoundError(f"Could not find a YOLO dataset root under {root}")


def upload_or_use_barn_zip(destination: Path = RAW_ROOT / "chicken-barn-images-upload") -> Path:
    clean_dir(destination)

    if BARN_DATASET_ZIP_PATH:
        zip_path = Path(BARN_DATASET_ZIP_PATH).expanduser().resolve()
    elif IN_COLAB:
        print("No Roboflow version found.")
        print("In Roboflow, click Export, choose YOLOv8, download the zip, then upload it here.")
        from google.colab import files
        uploaded = files.upload()
        if len(uploaded) != 1:
            raise ValueError("Upload exactly one YOLOv8 dataset zip.")
        zip_path = Path(next(iter(uploaded))).resolve()
    else:
        raise RuntimeError(
            "No Roboflow version found. Export the dataset from Roboflow as YOLOv8 zip "
            "and set BARN_DATASET_ZIP_PATH to that zip file."
        )

    with zipfile.ZipFile(zip_path) as archive:
        archive.extractall(destination)

    return find_yolo_dataset_root(destination)


def resolve_roboflow_project_version(project, preferred_version: int | None = None, max_scan: int = 100):
    if preferred_version is not None:
        return preferred_version, project.version(preferred_version)

    for version_number in range(max_scan, 0, -1):
        try:
            return version_number, project.version(version_number)
        except RuntimeError as exc:
            if "not found" not in str(exc).lower():
                raise

    return None, None


def download_barn_roboflow_dataset(destination: Path = RAW_ROOT / "chicken-barn-images") -> Path:
    from roboflow import Roboflow

    api_key = get_secret("ROBOFLOW_API_KEY")
    if not api_key:
        raise RuntimeError("Set ROBOFLOW_API_KEY as an environment variable or Colab Secret first.")

    print("loading Roboflow workspace...")
    rf = Roboflow(api_key=api_key)
    workspace = rf.workspace(BARN_ROBOFLOW_WORKSPACE)

    print("loading Roboflow project...")
    project = workspace.project(BARN_ROBOFLOW_PROJECT)

    version_number, version = resolve_roboflow_project_version(
        project,
        preferred_version=BARN_ROBOFLOW_VERSION,
    )

    if version is None:
        return upload_or_use_barn_zip()

    print(f"using Roboflow version {version_number}")
    clean_dir(destination)
    dataset = version.download("yolov8", location=str(destination), overwrite=True)

    return find_yolo_dataset_root(Path(dataset.location))


barn_raw_root = download_barn_roboflow_dataset()

barn_processed_root = process_yolo_source(
    source_root=barn_raw_root,
    output_root=PROCESSED_ROOT / "chicken_barn_images",
    class_map={
        "Chicken": "chicken",
        "chicken": "chicken",

        "Fox": "fox",
        "fox": "fox",
        "Dog": "dog",
        "dog": "dog",
        "Cat": "cat",
        "cat": "cat",

        "Person": "person",
        "person": "person",

        "Bird": "other_bird",
        "bird": "other_bird",
        "Other Bird": "other_bird",
        "other bird": "other_bird",
        "other_bird": "other_bird",

        "null": None,
        "Null": None,
        "NULL": None,
    },
    source_name="chicken_barn_images",
    keep_empty_images=True,
)
```

```python
# Check rebuild merged dataset from already-processed sources, remapping old class IDs to v4 schema.
from pathlib import Path
from collections import Counter
import json
import shutil
import yaml

CANONICAL_NAMES = [
    "chicken",
    "other_poultry",
    "rodent",
    "fox",
    "cat",
    "dog",
    "bird_of_prey",
    "other_bird",
    "person",
]
CANONICAL_ID = {name: index for index, name in enumerate(CANONICAL_NAMES)}

OLD_NAMES_WITH_MARTEN = [
    "chicken",
    "other_poultry",
    "rodent",
    "fox",
    "cat",
    "dog",
    "marten_weasel",
    "bird_of_prey",
    "other_bird",
    "person",
]

def write_data_yaml(dataset_root: Path, names: list[str] | None = None) -> None:
    names = list(names or CANONICAL_NAMES)
    data = {
        "path": str(Path(dataset_root).resolve()),
        "train": "train/images",
        "val": "valid/images",
        "test": "test/images",
        "nc": len(names),
        "names": names,
    }
    (Path(dataset_root) / "data.yaml").write_text(yaml.safe_dump(data, sort_keys=False), encoding="utf-8")

def read_data_yaml_names(dataset_root: Path) -> list[str]:
    data = yaml.safe_load((Path(dataset_root) / "data.yaml").read_text(encoding="utf-8"))
    names = data["names"]
    if isinstance(names, dict):
        return [str(names[key]) for key in sorted(names, key=lambda key: int(key))]
    return [str(name) for name in names]

def max_label_id(dataset_root: Path) -> int:
    max_id = -1
    for label_path in Path(dataset_root).glob("*/*/*.txt"):
        if label_path.parent.name != "labels":
            continue
        for line in label_path.read_text(encoding="utf-8").splitlines():
            parts = line.split()
            if not parts:
                continue
            try:
                max_id = max(max_id, int(float(parts[0])))
            except ValueError:
                pass
    return max_id

def schema_for_source(source_root: Path) -> list[str]:
    names = read_data_yaml_names(source_root)
    highest_id = max_label_id(source_root)

    if highest_id >= len(names) and highest_id < len(OLD_NAMES_WITH_MARTEN):
        print(f"{source_root.name}: data.yaml has {len(names)} names but labels use id {highest_id}; using old marten schema")
        return OLD_NAMES_WITH_MARTEN

    if "marten_weasel" in names:
        print(f"{source_root.name}: remapping old marten schema to current v4 schema")
    return names

def remap_label_lines(label_path: Path, source_names: list[str]) -> tuple[list[str], Counter]:
    output_lines = []
    counts = Counter()

    for line in label_path.read_text(encoding="utf-8").splitlines():
        parts = line.split()
        if len(parts) < 5:
            continue

        old_id = int(float(parts[0]))
        if old_id < 0 or old_id >= len(source_names):
            raise ValueError(f"{label_path}: class id {old_id} is outside source schema {source_names}")

        old_name = source_names[old_id]
        if old_name == "marten_weasel":
            continue
        if old_name not in CANONICAL_ID:
            raise ValueError(f"{label_path}: cannot map source class {old_name!r} into current v4 schema")

        new_name = old_name
        new_id = CANONICAL_ID[new_name]
        output_lines.append(" ".join([str(new_id), *parts[1:]]))
        counts[new_name] += 1

    return output_lines, counts

def clean_generated_dir(path: Path) -> None:
    path = Path(path).resolve()
    work_dir = Path(WORK_DIR).resolve()
    if path != work_dir and work_dir not in path.parents:
        raise ValueError(f"Refusing to clean outside WORK_DIR: {path}")
    if path.exists():
        shutil.rmtree(path)
    path.mkdir(parents=True, exist_ok=True)

def existing_dataset(value) -> Path | None:
    if value is None:
        return None

    if isinstance(value, Path):
        path = value
    elif isinstance(value, str):
        path = Path(value)
    else:
        return None

    if path.exists() and (path / "data.yaml").exists():
        return path

    return None


candidate_roots = []

for variable_name in ("poultry_processed_root", "open_images_processed_root", "barn_processed_root"):
    candidate_roots.append(globals().get(variable_name))

for variable_name in ("lila_processed_roots", "additional_processed_roots"):
    value = globals().get(variable_name, [])
    if isinstance(value, (list, tuple, set)):
        candidate_roots.extend(value)
    else:
        candidate_roots.append(value)

print("Using processed roots:")
for root in processed_roots:
    print(f"  - {root}")

candidate_roots = []
for variable_name in ("poultry_processed_root", "open_images_processed_root", "barn_processed_root"):
    candidate_roots.append(globals().get(variable_name))

candidate_roots.extend(globals().get("lila_processed_roots", []))
candidate_roots.extend(globals().get("additional_processed_roots", []))

processed_roots = []
seen_roots = set()
for candidate in candidate_roots:
    root = existing_dataset(candidate)
    if root is None:
        continue
    resolved = str(root.resolve())
    if resolved not in seen_roots:
        processed_roots.append(root)
        seen_roots.add(resolved)

if not processed_roots:
    raise RuntimeError("No processed source roots found. Run source processing cells first, not the download cells.")

def merge_canonical_yolo_sources_schema_safe(source_roots: list[Path], output_root: Path = FINAL_DATASET_ROOT) -> Path:
    output_root = Path(output_root)
    clean_generated_dir(output_root)

    for split in SPLITS:
        (output_root / split / "images").mkdir(parents=True, exist_ok=True)
        (output_root / split / "labels").mkdir(parents=True, exist_ok=True)

    manifest_rows = []
    merged_counts = {split: Counter() for split in SPLITS}

    for source_root in source_roots:
        source_root = Path(source_root)
        source_names = schema_for_source(source_root)
        source_prefix = slugify(source_root.name)

        manifest_by_image = {}
        manifest_path = source_root / "source_manifest_v4.jsonl"
        if manifest_path.exists():
            for line in manifest_path.read_text(encoding="utf-8").splitlines():
                if line.strip():
                    row = json.loads(line)
                    manifest_by_image[Path(row.get("image", "")).name] = row

        for split in SPLITS:
            src_image_dir = source_root / split / "images"
            src_label_dir = source_root / split / "labels"
            dst_image_dir = output_root / split / "images"
            dst_label_dir = output_root / split / "labels"

            for image_path in image_files(src_image_dir):
                label_path = src_label_dir / f"{image_path.stem}.txt"
                if not label_path.exists():
                    continue

                output_lines, counts = remap_label_lines(label_path, source_names)
                if not output_lines:
                    continue

                dest_stem = f"{source_prefix}_{slugify(image_path.stem)}"
                dest_image = copy_image(image_path, dst_image_dir, dest_stem)
                dest_label = dst_label_dir / f"{dest_stem}.txt"
                dest_label.write_text("\n".join(output_lines) + "\n", encoding="utf-8")

                source_row = dict(manifest_by_image.get(image_path.name, {}))
                source_row.update({
                    "image": str(dest_image.relative_to(output_root)).replace("\\", "/"),
                    "label": str(dest_label.relative_to(output_root)).replace("\\", "/"),
                    "split": split,
                    "merged_from": source_root.name,
                    "condition": source_row.get("condition", "day"),
                    "counts": dict(counts),
                })
                manifest_rows.append(source_row)
                merged_counts[split].update(counts)

    with (output_root / "source_manifest_v4.jsonl").open("w", encoding="utf-8") as manifest_file:
        for row in manifest_rows:
            manifest_file.write(json.dumps(row, sort_keys=True) + "\n")

    write_data_yaml(output_root)
    print(f"Schema-safe merged dataset written to {output_root}")
    for split in SPLITS:
        print(split, dict(merged_counts[split].most_common()))
    return output_root

final_dataset_root = merge_canonical_yolo_sources_schema_safe(processed_roots)
print_summary(summarize_yolo_source(final_dataset_root))
```

## Merge processed sources

```python
def merge_canonical_yolo_sources(source_roots: list[Path], output_root: Path = FINAL_DATASET_ROOT) -> Path:
    clean_dir(output_root)

    for split in SPLITS:
        (output_root / split / "images").mkdir(parents=True, exist_ok=True)
        (output_root / split / "labels").mkdir(parents=True, exist_ok=True)

    manifest_rows = []
    for source_root in source_roots:
        source_prefix = slugify(source_root.name)
        manifest_by_image = {}
        manifest_path = source_root / "source_manifest_v4.jsonl"
        if manifest_path.exists():
            for line in manifest_path.read_text(encoding="utf-8").splitlines():
                if not line.strip():
                    continue
                row = json.loads(line)
                manifest_by_image[Path(row.get("image", "")).name] = row

        for split in SPLITS:
            src_image_dir = source_root / split / "images"
            src_label_dir = source_root / split / "labels"
            dst_image_dir = output_root / split / "images"
            dst_label_dir = output_root / split / "labels"

            for image_path in image_files(src_image_dir):
                label_path = src_label_dir / f"{image_path.stem}.txt"
                if not label_path.exists():
                    continue
                dest_stem = f"{source_prefix}_{slugify(image_path.stem)}"
                dest_image = copy_image(image_path, dst_image_dir, dest_stem)
                dest_label = dst_label_dir / f"{dest_stem}.txt"
                shutil.copy2(label_path, dest_label)
                source_row = dict(manifest_by_image.get(image_path.name, {}))
                source_row.update({
                    "image": str(dest_image.relative_to(output_root)).replace("\\", "/"),
                    "label": str(dest_label.relative_to(output_root)).replace("\\", "/"),
                    "split": split,
                    "merged_from": source_root.name,
                    "condition": source_row.get("condition", "day"),
                })
                manifest_rows.append(source_row)

    with (output_root / "source_manifest_v4.jsonl").open("w", encoding="utf-8") as manifest_file:
        for row in manifest_rows:
            manifest_file.write(json.dumps(row, sort_keys=True) + "\n")

    write_data_yaml(output_root)
    return output_root

def valid_dataset_root(value) -> Path | None:
    if value is None:
        return None
    if not isinstance(value, (str, Path)):
        return None

    path = Path(value)
    if not path.exists():
        return None
    if not (path / "data.yaml").exists():
        return None
    return path


candidate_roots = [
    poultry_processed_root,
    open_images_processed_root,
    barn_processed_root,
    *lila_processed_roots,
    *additional_processed_roots,
]

processed_roots = [
    root
    for root in (valid_dataset_root(candidate) for candidate in candidate_roots)
    if root is not None
]

if processed_roots:
    final_dataset_root = merge_canonical_yolo_sources(processed_roots)
    print(f"Final dataset written to {final_dataset_root}")
    print_summary(summarize_yolo_source(final_dataset_root))
else:
    final_dataset_root = None
    print("No processed sources are available yet.")
```

## Balance final dataset distribution

This down-samples overrepresented classes after merging all sources. It intentionally caps `person`
and keeps predator classes near the same training count.


```python
BALANCED_DATASET_ROOT = WORK_DIR / "dataset_yolo_balanced"
SKIP_CANONICAL_CLASSES = {"marten_weasel"}
SCARCE_KEEP_CLASSES = {"chicken", "person", "bird_of_prey"}

TARGET_OBJECTS_BY_SPLIT = {
    "train": {
        "chicken": 10**9,
        "other_poultry": 1000,
        "rodent": 1100,
        "fox": 1200,
        "cat": 1100,
        "dog": 1200,
        "bird_of_prey": 10**9,
        "other_bird": 700,
        "person": 10**9,
    },
    "valid": {
        "chicken": 10**9,
        "other_poultry": 180,
        "rodent": 148,
        "fox": 220,
        "cat": 160,
        "dog": 200,
        "bird_of_prey": 10**9,
        "other_bird": 120,
        "person": 10**9,
    },
    "test": {
        "chicken": 10**9,
        "other_poultry": 180,
        "rodent": 160,
        "fox": 220,
        "cat": 160,
        "dog": 200,
        "bird_of_prey": 10**9,
        "other_bird": 120,
        "person": 10**9,
    },
}

def label_counts_for_file(label_path: Path) -> Counter:
    counts = Counter()
    for line_number, line in enumerate(label_path.read_text(encoding="utf-8").splitlines(), start=1):
        parsed = parse_yolo_line(line)
        if parsed is None:
            continue
        class_id, _box = parsed
        if not 0 <= class_id < len(CANONICAL_NAMES):
            raise ValueError(f"{label_path}:{line_number}: stale class id {class_id}; rerun remapping merge first.")
        class_name = CANONICAL_NAMES[class_id]
        if class_name not in SKIP_CANONICAL_CLASSES:
            counts[class_name] += 1
    return counts

def split_records(dataset_root: Path, split: str) -> list[dict]:
    manifest_by_image = load_manifest_by_image(dataset_root)
    records = []
    for image_path in image_files(dataset_root / split / "images"):
        label_path = dataset_root / split / "labels" / f"{image_path.stem}.txt"
        if not label_path.exists():
            continue
        counts = label_counts_for_file(label_path)
        if not counts:
            continue
        manifest = manifest_by_image.get(image_path.name, {})
        records.append({
            "image_path": image_path,
            "label_path": label_path,
            "counts": counts,
            "condition": manifest.get("condition", "day"),
            "manifest": manifest,
        })
    return records

def available_counts(records: list[dict]) -> Counter:
    counts = Counter()
    for record in records:
        counts.update(record["counts"])
    return counts

def effective_targets(records: list[dict], requested: dict[str, int]) -> dict[str, int]:
    available = available_counts(records)
    targets = {}
    for name in CANONICAL_NAMES:
        if name in SKIP_CANONICAL_CLASSES or available[name] <= 0:
            continue
        requested_target = requested.get(name, 0)
        targets[name] = available[name] if requested_target >= 10**9 else min(requested_target, available[name])
    return targets

def can_add(current: Counter, add: Counter, targets: dict[str, int]) -> bool:
    return all(name in targets and current[name] + value <= targets[name] for name, value in add.items())

def would_improve(record: dict, counts: Counter, targets: dict[str, int]) -> bool:
    return any(record["counts"].get(name, 0) and counts[name] < target for name, target in targets.items())

def score_record(record: dict, counts: Counter, targets: dict[str, int]) -> tuple[float, int]:
    score = 0.0
    for name, value in record["counts"].items():
        deficit = max(0, targets.get(name, 0) - counts[name])
        if deficit:
            weight = 5.0 if name in SCARCE_KEEP_CLASSES else 1.0
            score += min(value, deficit) * weight
    return score, -sum(record["counts"].values())

def copy_record(record: dict, dst_image_dir: Path, dst_label_dir: Path, prefix: str) -> tuple[Path, Path]:
    dest_stem = f"{prefix}_{slugify(record['image_path'].stem)}"
    dest_image = copy_image(record["image_path"], dst_image_dir, dest_stem)
    dest_label = dst_label_dir / f"{dest_stem}.txt"
    shutil.copy2(record["label_path"], dest_label)
    return dest_image, dest_label

def balance_canonical_yolo_dataset(input_root: Path, output_root: Path = BALANCED_DATASET_ROOT) -> Path:
    if load_class_names(input_root) != CANONICAL_NAMES:
        raise ValueError("Merged data.yaml names do not match active CANONICAL_NAMES.")

    clean_dir(output_root)
    manifest_rows = []

    for split in SPLITS:
        dst_image_dir = output_root / split / "images"
        dst_label_dir = output_root / split / "labels"
        dst_image_dir.mkdir(parents=True, exist_ok=True)
        dst_label_dir.mkdir(parents=True, exist_ok=True)

        records = split_records(input_root, split)
        targets = effective_targets(records, TARGET_OBJECTS_BY_SPLIT[split])
        selected = []
        selected_paths = set()
        counts = Counter()

        while True:
            candidates = [
                record for record in records
                if str(record["image_path"]) not in selected_paths
                and would_improve(record, counts, targets)
                and can_add(counts, record["counts"], targets)
            ]
            if not candidates:
                break

            best = max(candidates, key=lambda record: score_record(record, counts, targets))
            if score_record(best, counts, targets)[0] <= 0:
                break

            selected.append(best)
            selected_paths.add(str(best["image_path"]))
            counts.update(best["counts"])

        for index, record in enumerate(selected):
            dest_image, dest_label = copy_record(record, dst_image_dir, dst_label_dir, prefix=f"{split}_{index:06d}")
            row = dict(record.get("manifest", {}))
            row.update({
                "image": str(dest_image.relative_to(output_root)).replace("\\", "/"),
                "label": str(dest_label.relative_to(output_root)).replace("\\", "/"),
                "split": split,
                "condition": record["condition"],
                "counts": dict(record["counts"]),
            })
            manifest_rows.append(row)

        print(f"\nBalanced {split}: selected {len(selected)} / {len(records)} images")
        for name in CANONICAL_NAMES:
            target = targets.get(name, 0)
            status = "OK" if target == 0 or counts[name] >= target else "LOW"
            print(f"  - {name}: {counts[name]} / target {target} [{status}]")

    write_data_yaml(output_root, CANONICAL_NAMES)
    with (output_root / "source_manifest_v4.jsonl").open("w", encoding="utf-8") as manifest_file:
        for row in manifest_rows:
            manifest_file.write(json.dumps(row, sort_keys=True) + "\n")
    return output_root

final_dataset_root = balance_canonical_yolo_dataset(final_dataset_root)
print_summary(summarize_yolo_source(final_dataset_root))
```

## Validate final YOLO dataset

```python
def validate_yolo_dataset(dataset_root: Path) -> list[str]:
    errors: list[str] = []
    names = load_class_names(dataset_root)
    if names != CANONICAL_NAMES:
        errors.append("data.yaml names do not match CANONICAL_NAMES")

    for split in SPLITS:
        image_dir = dataset_root / split / "images"
        label_dir = dataset_root / split / "labels"
        image_stems = {path.stem for path in image_files(image_dir)}
        label_stems = {path.stem for path in label_dir.glob("*.txt")} if label_dir.exists() else set()

        missing_labels = image_stems - label_stems
        orphan_labels = label_stems - image_stems
        if missing_labels:
            errors.append(f"{split}: {len(missing_labels)} images have no label file")
        if orphan_labels:
            errors.append(f"{split}: {len(orphan_labels)} label files have no image")

        for label_path in sorted(label_dir.glob("*.txt")) if label_dir.exists() else []:
            for line_number, line in enumerate(label_path.read_text(encoding="utf-8").splitlines(), start=1):
                if not line.strip():
                    continue
                parsed = parse_yolo_line(line)
                if parsed is None:
                    errors.append(f"{label_path}:{line_number}: invalid YOLO row")
                    continue
                class_id, box = parsed
                if not 0 <= class_id < len(CANONICAL_NAMES):
                    errors.append(f"{label_path}:{line_number}: class id {class_id} out of range")
                if any(value < 0 or value > 1 for value in box):
                    errors.append(f"{label_path}:{line_number}: box value outside 0..1")

    return errors


if final_dataset_root is not None:
    validation_errors = validate_yolo_dataset(final_dataset_root)
    if validation_errors:
        print("Validation failed:")
        for error in validation_errors[:50]:
            print(f"  - {error}")
        if len(validation_errors) > 50:
            print(f"  ... {len(validation_errors) - 50} more errors")
    else:
        print("Validation passed.")
else:
    print("No final dataset to validate yet.")
```

## Visual quality checks

Run this after creating `final_dataset_root`. Inspect samples before training. In particular, verify that predators are labeled where visible, irrelevant large-bird labels are gone, and poultry polygons became reasonable boxes.


```python
def draw_yolo_boxes(image_path: Path, label_path: Path, names: list[str]) -> Image.Image:
    image = Image.open(image_path).convert("RGB")
    width, height = image.size
    draw = ImageDraw.Draw(image)

    if label_path.exists():
        for line in label_path.read_text(encoding="utf-8").splitlines():
            parsed = parse_yolo_line(line)
            if parsed is None:
                continue
            class_id, (x_center, y_center, box_width, box_height) = parsed
            x_min = (x_center - box_width / 2) * width
            y_min = (y_center - box_height / 2) * height
            x_max = (x_center + box_width / 2) * width
            y_max = (y_center + box_height / 2) * height
            label = names[class_id] if 0 <= class_id < len(names) else str(class_id)
            draw.rectangle([x_min, y_min, x_max, y_max], outline="red", width=3)
            draw.text((x_min + 3, y_min + 3), label, fill="red")
    return image


def show_random_samples(dataset_root: Path, split: str = "train", count: int = 9, seed: int = 7) -> None:
    import random

    random.seed(seed)
    names = load_class_names(dataset_root)
    images = image_files(dataset_root / split / "images")
    if not images:
        print(f"No images found for {split}")
        return

    selected = random.sample(images, k=min(count, len(images)))
    columns = min(3, len(selected))
    rows = (len(selected) + columns - 1) // columns
    plt.figure(figsize=(5 * columns, 4 * rows))

    for index, image_path in enumerate(selected, start=1):
        label_path = dataset_root / split / "labels" / f"{image_path.stem}.txt"
        rendered = draw_yolo_boxes(image_path, label_path, names)
        plt.subplot(rows, columns, index)
        plt.imshow(rendered)
        plt.axis("off")
        plt.title(image_path.name[:40])

    plt.tight_layout()
    plt.show()


if final_dataset_root is not None:
    show_random_samples(final_dataset_root, split="train", count=9)
else:
    print("No final dataset to visualize yet.")
```

## Class balance

```python
def plot_class_balance(dataset_root: Path) -> None:
    summary = summarize_yolo_source(dataset_root)
    names = load_class_names(dataset_root)
    split_counts = {split: [summary[split][name] for name in names] for split in SPLITS}

    x_positions = range(len(names))
    bottom = [0] * len(names)
    plt.figure(figsize=(14, 5))
    for split in SPLITS:
        values = split_counts[split]
        plt.bar(x_positions, values, bottom=bottom, label=split)
        bottom = [current + value for current, value in zip(bottom, values)]

    plt.xticks(list(x_positions), names, rotation=45, ha="right")
    plt.ylabel("objects")
    plt.title("Canonical class balance")
    plt.legend()
    plt.tight_layout()
    plt.show()


if final_dataset_root is not None:
    plot_class_balance(final_dataset_root)
else:
    print("No final dataset to plot yet.")
```
